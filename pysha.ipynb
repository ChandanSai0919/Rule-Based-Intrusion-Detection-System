```python
import pyshark
import asyncio
import nest_asyncio

# Apply nest_asyncio to handle nested event loops
nest_asyncio.apply()

INTERFACE = "\\Device\\NPF_{1875F0DF-387A-4CF9-8679-7FB093D7C5E7}"  # Correct interface

async def capture_packets():
    """Capture packets asynchronously without using 'async for'."""
    capture = pyshark.LiveCapture(interface=INTERFACE)
    print("✅ Listening for packets...")

    # Run sniff_continuously() synchronously but asynchronously wrapped
    loop = asyncio.get_event_loop()
    await loop.run_in_executor(None, sniff_and_print, capture)

def sniff_and_print(capture):
    """Helper function to sniff packets synchronously and print them."""
    for packet in capture.sniff_continuously(packet_count=5):
        print(packet)

# Run the capture using asyncio
if __name__ == "__main__":
    loop = asyncio.get_event_loop()
    loop.run_until_complete(capture_packets())
```
